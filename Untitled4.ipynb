1.. What is the difference between a function and a method in Python?
 - In Python, both methods and functions are used to perform specific tasks, but they have distinct differences in their usage and implementation.

Functions in Python

A function is a block of code that performs a specific task and can be called by its name. Functions are defined using the def keyword and can have parameters and return values. Functions can be categorized into three types:

1. Built-in Functions: These are pre-defined functions in Python, such as print(), sum(), etc.

2. User-Defined Functions (UDFs): These are functions defined by the user to perform specific tasks.

3. Anonymous Functions: Also known as lambda functions, these are functions without a name and are defined using the lambda keyword.

Example of a User-Defined Function

def add(num1, num2):
return num1 + num2

print(add(1, 2)) # Output: 3
In this example, add is a user-defined function that takes two parameters and returns their sum.

Methods in Python

A method is a function that is associated with an object. Methods are defined within a class and can access and modify the attributes of the class. Methods always have self as their first parameter, which refers to the instance of the class.

Example of a Method

class Addition:
def add(self, num1, num2):
return num1 + num2

addition1 = Addition() # Object Instantiation
print(addition1.add(2, 3)) # Output: 5
In this example, add is a method defined within the Addition class and is called using an instance of the class
.

Key Differences Between Methods and Functions

1. Location: Functions are defined outside of classes, while methods are defined inside classes.

2. Association: Functions are independent and not associated with any objects, whereas methods are associated with the objects of the class they are defined in.

3. Calling: Functions can be called by their name alone, while methods need to be called using an object or class name followed by a dot operator.

4. Parameters: Methods always have self as their first parameter, while functions do not require self.

5. Access to Class Attributes: Methods can access and modify the attributes of the class, while functions cannot.

In Python, function arguments are values passed to a function when it is called. These arguments allow functions to operate on different data inputs. There are several types of function arguments in Python, each serving a unique purpose.

Positional arguments are the most common type of arguments. They are passed to the function in the same order as the parameters are defined. For example:

Keyword arguments are passed to the function by explicitly specifying the parameter name along with its value. This allows the arguments to be passed in any order. For example:

Default arguments are parameters that assume a default value if no argument is provided during the function call. For example:

In this example, the message parameter has a default value of "Hello"

Variable-Length Arguments

Variable-length arguments allow a function to accept an arbitrary number of arguments. There are two types of variable-length arguments: *args and **kwargs.

*args

The *args syntax allows a function to accept any number of positional arguments, which are stored in a tuple. For example:

In this example, *args collects all the positional arguments into a tuple
3

**kwargs

In Python, iterators and iterables are fundamental concepts that are often used in data processing and iteration. Understanding the difference between them is crucial for writing efficient and readable code.

Iterable

An iterable is any Python object capable of returning its members one at a time, allowing it to be looped over in a for-loop. Examples of iterables include lists, tuples, dictionaries, sets, and strings. An object is considered iterable if it implements the __iter__() method, which returns an iterator object. For example:

# Example of an iterable
cities = ["Berlin", "Vienna", "Zurich"]
for city in cities:
print(city)
In this example, cities is an iterable because it can be looped over in a for-loop.

Iterator

An iterator is an object that represents a stream of data. It is used to iterate over an iterable object using the __next__() method. An iterator keeps track of the current position during iteration and returns the next item when __next__() is called. If there are no more items, it raises a StopIteration exception. For example:

# Example of an iterator
cities = ["Berlin", "Vienna", "Zurich"]
iterator_obj = iter(cities)
print(next(iterator_obj)) # Output: Berlin
print(next(iterator_obj)) # Output: Vienna
print(next(iterator_obj)) # Output: Zurich
In this example, iterator_obj is an iterator created from the iterable cities
1
.

Key Differences

Definition: Iterable: An object that can be iterated over. Iterator: An object that performs the iteration over an iterable.

Methods: Iterable: Must implement the __iter__() method. Iterator: Must implement both __iter__() and __next__() methods.

State: Iterable: Does not store the iteration state. Iterator: Stores the iteration state.

Usage: Iterable: Can be used in a for-loop directly. Iterator: Can be used with the next() function to get the next item.

Multiple Iterations: Iterable: Can be iterated over multiple times. Iterator: Can be iterated over only once unless explicitly reset.

Practical Example

Here is a practical example demonstrating the difference between an iterable and an iterator:

# Iterable example
class MyIterable:
def __init__(self, data):
self.data = data

def __iter__(self):
return MyIterator(self.data)

# Iterator example
class MyIterator:
def __init__(self, data):
self.data = data
self.index = 0

def __iter__(self):
return self

def __next__(self):
if self.index < len(self.data):
item = self.data[self.index]
self.index += 1
return item
else:
raise StopIteration

# Using the iterable and iterator
my_iterable = MyIterable([1, 2, 3])
for item in my_iterable:
print(item)
In this example, MyIterable is an iterable class that returns an instance of MyIterator when __iter__() is called. MyIterator is an iterator class that implements the __next__() method to return the next item in the data

Generators in Python are a special type of function that return an iterator, which can be iterated over one value at a time. They are particularly useful when dealing with large datasets or streams of data, as they allow for lazy evaluation, meaning values are generated on-the-fly and not stored in memory all at once

Creating Generators

To create a generator in Python, you define a function using the def keyword, but instead of using return to return a value, you use the yield keyword. This makes the function a generator function. When called, it returns a generator object that can be iterated over to produce the values.

Example of a Generator Function

def my_generator(n):
value = 0
while value < n:
yield value
value += 1

# Using the generator
for value in my_generator(3):
print(value)
Output:

0
1
2
In this example, the my_generator function yields values from 0 to n-1. Each time yield is called, the function's state is saved, and it resumes from there the next time it is called
2
.

Generator Expressions

Generator expressions provide a concise way to create generators. They are similar to list comprehensions but use parentheses instead of square brackets.

Example of a Generator Expression

squares_generator = (i * i for i in range(5))

for i in squares_generator:
print(i)
Output:

0
1
4
9
16
This generator expression creates a generator that produces the squares of numbers from 0 to 4
2
.

Advantages of Generators

Memory Efficiency: Generators generate values on-the-fly, reducing memory usage compared to storing the entire sequence in memory
3
.

Lazy Evaluation: Values are computed only when needed, which can lead to performance improvements, especially with large datasets
3
.

Maintain State: Generators automatically maintain their state between yield statements, making them suitable for iterative processes
3
.

Convenience: Generators simplify code for iterating over large datasets or streams of data without needing to manage state explicitly
3
.

Advanced Generator Methods

Generators also support advanced methods like .send(), .throw(), and .close().

Using .send()

The .send() method allows you to send a value back to the generator.

def infinite_palindromes():

while True:
if str(num) == str(num)[::-1]:
i = (yield num)
if i is not None:
num = i

In this example, the generator checks for palindromes and can receive a new starting point via .send()
1
.

Using .throw()

The .throw() method allows you to throw an exception inside the generator.

This will raise a ValueError inside the generator
1
.

Using .close()

The .close() method stops the generator.

```python
11. Imagine an accounting routine used in a book shop. It works on a list with sublists, which look like this:
 Write a Python program, which returns a list with 2-tuples. Each tuple consists of the order number and the
product of the price per item and the quantity. The product should be increased by 10,- € if the value of the
order is smaller than 100,00 €.
 Write a Python program using lambda and map.


def process_orders(orders):
    """
    Processes a list of bookshop orders, calculating the total price and adding a surcharge if necessary.

    Args:
        orders (list): A list of orders, where each order is a list containing:
                       [order_number, price_per_item, quantity]

    Returns:
        list: A list of 2-tuples, where each tuple contains the order number and the
              final price (price_per_item * quantity + 10 if the total is less than 100).
    """
    processed_orders = list(map(lambda order: (order[0], order[1] * order[2] + (10 if order[1] * order[2] < 100 else 0)), orders))
    return processed_orders

# Example Usage:
orders = [
    [121, 8, 11],
    [122, 11, 1],
    [123, 12, 2]
]

processed_orders = process_orders(orders)
print(processed_orders)

In this program:


The process_orders function takes a list of orders as input. Each order is a list containing the order number, price per item, and quantity
.
A lambda function is used within the map function to calculate the total price for each order
. If the order value is less than 100, a 10 € surcharge is added
.
The map function applies this lambda function to each order in the list, generating a new list of 2-tuples containing the order number and the calculated price


.
The result is converted to a list and returned
. Lambda functions are often used with map() for concise, anonymous function definitions

```
